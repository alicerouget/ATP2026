### TPC2: Adivinha o número
- Crie um programa em Python para jogar o jogo "Adivinha o número":
    - O jogo pode ter 2 modalidades: computador pensa num número (entre 0 e 100), utilizador tenta adivinhar; ou, o utilizador pensa num número (entre 0 e 100) e o computador tenta adivinhar;
    - Quem tenta adivinhar responde com uma das afirmações: "Acertou", "O número que pensei é Maior" ou "O número que pensei é Menor";
    - Uma vez descoberto o número o programa deverá terminar imprimindo o número de tentativas que quem adivinhou usou para chegar ao resultado.

```python
# Solução do TPC 2

print("1- Adivinha o número do computador")
print("2- O computador adivinha o teu número")
modalidade = int(input("Escolha a modalidade 1 ou 2"))

import random

if modalidade ==1:
    numerocomputador = random.randint(0, 100)
    tentativas = 0
    numerointroduzido = -1

    while numerointroduzido != numerocomputador:
        numerointroduzido = int(input("Adivinha o número do computador: "))
        tentativas = tentativas +1
        if numerointroduzido < numerocomputador:
            print("Ups, o número que pensei é maior!")
        if numerointroduzido > numerocomputador:
            print("Ups, o número que pensei é menor!")
    print("Acertou")
    print("Tentativas:", tentativas)


if modalidade ==2:
    print("Pensa num número de 0 a 100")
    intervalo = [0, 100]
    tentativas = 0
    respostautilizador = ""
    numerointroduzido = -1

    while respostautilizador != "Acertou":
        numerointroduzido = random.randint(intervalo[0], intervalo[1])
        tentativas = tentativas + 1
        print("O computador tenta:...", numerointroduzido)
        respostautilizador = input("Escreva (Acertou / Maior / Menor):")
        if respostautilizador =="Maior":
            intervalo[0] = numerointroduzido + 1
        if respostautilizador == "Menor":
            intervalo[1] = numerointroduzido - 1
print("Tentativas:", tentativas)
```
